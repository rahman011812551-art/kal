```python
%matplotlib widget

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

x = np.array([1,2,2,1,1])
y = np.array([1,1,2,2,1])

fig, ax = plt.subplots()

ax.set_xlim(-1,3)
ax.set_ylim(-1,3)
ax.grid()

line, = ax.plot([], [], linewidth=3)

def update(frame):

    skala = 1 - frame/100
    y_baru = y * skala

    line.set_data(x, y_baru)

    return line,

anim = FuncAnimation(
    fig,
    update,
    frames=100,
    interval=50
)

plt.show()
```
